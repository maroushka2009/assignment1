```python
#List
students_list=["Asha","Rahul","Priya"]
print("Original list :",students_list)
```

```python
students_list.append("Amit")
print("After append :",students_list)
```

```python
students_list.insert(1,"Neha")
print("After insert :",students_list)
```

```python
students_list[0]= "Ash"
print("After update :",students_list)
```

```python
students_list.remove("Rahul")
print("After removal :",students_list)
```

```python
students_list.pop(1)
print("After popping :",students_list)
```

```python
#Tuple
students_tuple=("Asha","Rahul","Priya")
print("Original tuple :",students_tuple)

```

```python
students_tuple = students_tuple + ("Amit",)
print("After concatenation :",students_tuple)
```

```python
temp=list(students_tuple)
temp[1]=("Rohan")
students_tuple= tuple(temp)
print("After update :",students_tuple)
```

```python
temp=list(students_tuple)
temp.remove("Priya")
students_tuple= tuple(temp)
print("After deleting :",students_tuple)
```

```python
print("len()",len(students_tuple),
      "| count('Asha')=", students_tuple.count("Asha"),
      "| index('Amit')=", students_tuple.index("Amit"))
```

```python
#Dictionary
students_dict={1:"Asha", 2:"Rahul", 3:"Priya"}
print("Original dict :",students_dict)
```

```python
students_dict[4]="Amit"
print("After add :", students_dict)
```

```python
students_dict.update({2:"Rohan", 5:"Neha"})
print("After update :",students_dict)
```

```python
students_dict.pop(3)
print("After pop(3) :",students_dict)
del students_dict[5]
print("After del [5] :", students_dict)
```

```python
print("get(1)=", students_dict.get(1), "| get(99)=", students_dict.get(99))
print("keys() :", list(students_dict.keys()))
print("items() :", list(students_dict.items()))
      
```
